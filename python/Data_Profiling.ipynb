Renaming Columns

DateTime Conversion

```python
fraud_data['trans_date_trans_time'] = pd.to_datetime(fraud_data['trans_date_trans_time'])
```

```python
fraud_data['transaction_date'] = fraud_data['trans_date_trans_time'].dt.date
fraud_data['transaction_hour'] = fraud_data['trans_date_trans_time'].dt.hour
fraud_data['transaction_day'] = fraud_data['trans_date_trans_time'].dt.day_name()
fraud_data['transaction_month'] = fraud_data['trans_date_trans_time'].dt.month_name()
```

```python
month_order = ['January','February','March','April','May','June',
               'July','August','September','October','November','December']

fraud_data['transaction_month'] = pd.Categorical(
    fraud_data['transaction_month'],
    categories=month_order,
    ordered=True
)
fraud_trend = fraud_data.groupby('transaction_month')['is_fraud'].count()
```

Creating Full Name column

```python
fraud_data['full_name'] = fraud_data['first_name'] + ' ' + fraud_data['last_name']
```

Creating Customer Age

```python
fraud_data['date_of_birth'] = pd.to_datetime(fraud_data['date_of_birth'])
today = pd.to_datetime('today')
fraud_data['customer_age'] = (today - fraud_data['date_of_birth']).dt.days // 365
```

Visualizations

Fraud Count

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.countplot(x='is_fraud', data=fraud_data)
plt.title('Distribution of Fraudulent vs Non-Fraudulent Transactions')
plt.xlabel('Is Fraud')
plt.ylabel('Count')
plt.show()
```

Transaction Amount Analysis

```python
plt.figure(figsize=(10,6))
sns.boxplot(x=fraud_data['amount'])
plt.xlim(0, 5000)
plt.title("Transaction Amount Distribution (Focused View)", fontsize=14)
plt.xlabel("Transaction Amount")

plt.show()
```

Plot Line chart

```python
plt.figure(figsize=(12,6))
fraud_trend.plot(marker='o')
plt.title("Fraud Transaction Trend by Month", fontsize=14)
plt.xlabel("Month")
plt.ylabel("Fraud Transactions")
plt.grid(True)
plt.show()
```

Fraud by Category

```python
fraud_by_category = fraud_data[fraud_data['is_fraud']==1]['category'].value_counts()

fraud_by_category.plot(kind='bar')
plt.title('Distribution of Fraudulent Transactions by Category')
plt.xlabel('Category')
plt.ylabel('Count') 
plt.show()
```

Fraud by Hour

```python
sns.countplot(x='transaction_hour', hue='is_fraud', data=fraud_data)
plt.title('Distribution of Transactions by Hour')
plt.xlabel('Transaction Hour')
plt.ylabel('Count')
plt.show()
```

Fraud by Day

```python
fraud_by_day = fraud_data[fraud_data['is_fraud']==1]['transaction_day'].value_counts()

fraud_by_day.plot(kind='bar')
plt.title('Distribution of Fraudulent Transactions by Day')
plt.xlabel('Transaction Day')
plt.ylabel('Count') 
plt.show()
```
